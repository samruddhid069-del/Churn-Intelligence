# Churn Analysis Notebook
Explore the dataset and check data quality before modelling.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

DATA_PATH = Path('..') / 'data' / 'exported_churn_data.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.isna().sum().sort_values(ascending=False))
```

```python
if 'churn_flag' in df.columns:
    churn = pd.to_numeric(df['churn_flag'], errors='coerce')
    print(f'Churn rate: {churn.mean() * 100:.2f}%')
```

```python
if {'plan_type', 'churn_flag'}.issubset(df.columns):
    data = df.dropna(subset=['plan_type', 'churn_flag']).groupby('plan_type')['churn_flag'].mean().mul(100).reset_index(name='Churn Rate (%)')
    fig = px.bar(data, x='plan_type', y='Churn Rate (%)', color='Churn Rate (%)', template='plotly_dark', title='Churn Rate by Plan')
    fig.show()
```
